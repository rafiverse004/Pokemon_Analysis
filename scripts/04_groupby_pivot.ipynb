```python
import pandas as pd
```

```python
df = pd.read_csv('../data/processed/02_cleaning.csv')

print(df)
```

```python

type_avg_total = (
    df.groupby('Type_1')['Total'].mean().sort_values(ascending = False)
)

print(type_avg_total)

type_avg_total.to_csv('../data/processed/Type_avg_total.csv', index = False)
```

```python
avg_attack_type = (
    df.groupby('Type_1')['Attack'].mean().sort_values(ascending = False)
)

print(avg_attack_type)

avg_attack_type.to_csv('../data/processed/Avg_attack_type.csv', index = False)
```

```python
avg_defense_type = (
    df.groupby('Type_1')['Defense'].mean().sort_values(ascending = False)
)

print(avg_defense_type)

avg_defense_type.to_csv('../data/processed/Avg_defense_type.csv', index = False)
```

```python
legendary_comparison = (
    df.groupby("Legendary")["Total"]
      .mean()
)

print(legendary_comparison)

legendary_comparison.to_csv('../data/processed/Legendary_comparison.csv', index = False)
```

```python
generation_avg_total = (
    df.groupby('Generation')['Total'].mean().sort_values(ascending = False)
)

print(generation_avg_total)

generation_avg_total.to_csv('../data/processed/Generation_avg_total.csv', index = False)
```

Key Findings

- Dragon type has the highest average Attack.

- Steel type has the highest average Defense.

- Dragon, Steel, and Flying rank highest by average Total.

- Legendary Pokémon have significantly higher average Total stats than non-legendary Pokémon.

- No consistent increase in average Total stats across generations.
