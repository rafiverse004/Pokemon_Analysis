```python
import pandas as pd
```

```python
df = pd.read_csv('../data/processed/02_cleaning.csv')

print(df)
```

```python
df['Offensive_Power'] = df['Attack'] + df['Sp_Atk']

df[["Name", "Attack", "Sp_Atk", "Offensive_Power"]].head()
```

```python
df["Defensive_Power"] = df["Defense"] + df["Sp_Def"]

df[["Name", "Defense", "Sp_Def", "Defensive_Power"]].head()
```

```python
df["Power_Diff"] = df["Offensive_Power"] - df["Defensive_Power"]

df[["Name", "Offensive_Power", "Defensive_Power", "Power_Diff"]].head()
```

💡 How to read it

- Power_Diff > 0 → offensive Pokémon
- Power_Diff < 0 → defensive Pokémon
- Power_Diff ≈ 0 → balanced Pokémon

```python
def classify_role(row):
    if row["Power_Diff"] > 20:
        return "Attacker"
    elif row["Power_Diff"] < -20:
        return "Defender"
    else:
        return "Balanced"

df["Role"] = df.apply(classify_role, axis=1)

df[["Name", "Offensive_Power", "Defensive_Power", "Role"]].head()
```
