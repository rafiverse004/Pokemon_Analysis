# 📊 Final Report

## 🎯 Project Summary
This project analyzed a Pokémon dataset using Python and Pandas to understand patterns in stats, types, generations, and gameplay roles.

The main goal was to practice real-world data analysis by cleaning raw data, exploring relationships, creating new features, and extracting insights.

---

## 🔍 Key Insights

### ⚔️ Pokémon Strength
- Dragon-type Pokémon have the highest average Attack.
- Steel-type Pokémon have the highest average Defense.
- Flying and Dragon types rank among the strongest overall in Total stats.

---

### ⭐ Legendary vs Non-Legendary
- Legendary Pokémon are significantly stronger on average compared to non-legendary Pokémon.
- However, they represent a very small portion of the dataset (~8%), making them rare high-power units.

---

### 📈 Generations
- Generation 4 has the highest average Total stats.
- Generation 2 has the lowest.
- There is no consistent upward trend in strength across generations.

---

### 🧪 Feature Engineering Insights
New features helped simplify Pokémon classification:

- Offensive Power = Attack + Special Attack
- Defensive Power = Defense + Special Defense
- Power Difference helped classify Pokémon into roles

### Pokémon Roles:
- Balanced Pokémon are the majority.
- Attackers and Defenders are fewer but more specialized.

---

## 🧠 Key Takeaways
- Pokémon stats are highly influenced by type and role.
- Legendary status strongly impacts overall strength.
- Most Pokémon are designed to be balanced rather than extreme attackers or defenders.
- Feature engineering made raw stats easier to interpret for gameplay analysis.
