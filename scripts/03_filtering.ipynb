```python
import pandas as pd
```

```python
df = pd.read_csv('../data/processed/02_cleaning.csv')
```

```python
df.columns
```

```python
legendary = df[df['Legendary'] == True]

print(legendary.shape)
print(legendary.head())

legendary.to_csv("../data/processed/legendary.csv", index=False)
```

```python
legendary_strong = df[(df['Legendary'] == True) & (df['Total'] >= 600)]

print(legendary_strong.shape)
print(legendary_strong.head())

legendary_strong.to_csv("../data/processed/legendary_strong.csv", index=False)
```

```python
non_legendary = df[df['Legendary'] == False]

print(non_legendary.shape)
print(non_legendary.head())

non_legendary.to_csv("../data/processed/non_legendary.csv", index=False)
```

```python
non_legendary_strong = df[(df['Legendary'] == False) & (df['Total'] >= 600)]

print(non_legendary_strong.shape)
print(non_legendary_strong.head())

non_legendary_strong.to_csv("../data/processed/non_legendary_strong.csv", index=False)
```

```python
fast = df[df["Speed"] > 100]

print(fast.shape)
print(fast.head())

fast.to_csv('../data/processed/fast.csv', index = False)
```

```python
water = df[df["Type_1"] == "Water"]

print(water.shape)
print(water.head())

water.to_csv('../data/processed/water.csv', index = False)
```

```python
fire_electric = df[
    (df["Type_1"] == "Fire") |
    (df["Type_1"] == "Electric")
]

print(fire_electric.shape)
print(fire_electric.head())

fire_electric.to_csv('../data/processed/fire_electric.csv', index = False)
```

```python
water_flying = df[
    (df["Type_1"] == "Water") &
    (df["Type_2"] == "Flying")
]

print(water_flying.shape)
print(water_flying.head())

water_flying.to_csv('../data/processed/water_flying.csv', index = False)
```
